Think of the dots as **compressed descriptions of the world** — not images or words directly, but “features” the model has learned.

A JEPA-style model roughly does this:

```text
current observation → encoder → latent dot
model predicts where the next latent dot should be
```

So in the plot:

- **colored dots** = hidden/latent states over time
- **blue rings** = selected current states
- **pink arrows** = “where the model thinks the future state will be”
- **spiral-ish path** = the world changing over time

The key idea: **it predicts the future in representation space**, not by drawing the future pixels.
